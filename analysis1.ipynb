```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Daten einlesen und ungültige Werte entfernen
times = top_data["TimeSinceZero"].dropna().to_numpy()

# Histogramm berechnen
counts, bin_edges = np.histogram(
    times,
    bins=100,
    range=(0, 10)
)

# Mittelpunkte der Histogramm-Bins
bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])


def scintillation_model(
    t,
    amplitude,
    t0,
    t_rise,
    t_fall,
    background
):
    """
    Modell:
        A * (1 - exp(-(t-t0)/t_rise))
          * exp(-(t-t0)/t_fall) + B

    Das Signal beginnt erst bei t >= t0.
    """
    dt = t - t0

    signal = np.zeros_like(t, dtype=float)
    mask = dt >= 0

    signal[mask] = (
        amplitude
        * (1.0 - np.exp(-dt[mask] / t_rise))
        * np.exp(-dt[mask] / t_fall)
    )

    return signal + background


# Poisson-Unsicherheit der Zählwerte
# Für leere Bins wird sigma=1 verwendet.
sigma = np.sqrt(np.maximum(counts, 1))

# Startwerte:
# amplitude, t0, t_rise, t_fall, background
initial_values = [
    counts.max(),
    0.0,
    0.9,
    2.1,
    0.0
]

# Untere und obere Parametergrenzen
bounds = (
    [0.0,    -1.0, 0.01, 0.01, 0.0],
    [np.inf,  2.0, 10.0, 20.0, np.inf]
)

# Fit durchführen
fit_parameters, covariance = curve_fit(
    scintillation_model,
    bin_centers,
    counts,
    p0=initial_values,
    sigma=sigma,
    absolute_sigma=True,
    bounds=bounds,
    maxfev=20_000
)

# Gefittete Parameter
amplitude, t0, t_rise, t_fall, background = fit_parameters

# Standardabweichungen der Fitparameter
parameter_errors = np.sqrt(np.diag(covariance))

print("Fit-Ergebnisse:")
print(
    f"Amplitude  = {amplitude:.4g} "
    f"± {parameter_errors[0]:.2g}"
)
print(
    f"t0         = {t0:.4f} "
    f"± {parameter_errors[1]:.4f} ns"
)
print(
    f"t_rise     = {t_rise:.4f} "
    f"± {parameter_errors[2]:.4f} ns"
)
print(
    f"t_fall     = {t_fall:.4f} "
    f"± {parameter_errors[3]:.4f} ns"
)
print(
    f"Background = {background:.4g} "
    f"± {parameter_errors[4]:.2g}"
)

# Glatte x-Achse für die Fitkurve
t_plot = np.linspace(0, 10, 1000)
fit_curve = scintillation_model(t_plot, *fit_parameters)

# Histogramm und Fit darstellen
plt.figure(figsize=(8, 5))

plt.hist(
    times,
    bins=100,
    range=(0, 10),
    histtype="step",
    linewidth=2,
    label="Top"
)

plt.plot(
    t_plot,
    fit_curve,
    color="tab:red",
    linewidth=2,
    label=(
        rf"Fit: $t_\mathrm{{rise}}="
        rf"{t_rise:.2f}\pm{parameter_errors[2]:.2f}$ ns"
        "\n"
        rf"$t_\mathrm{{fall}}="
        rf"{t_fall:.2f}\pm{parameter_errors[3]:.2f}$ ns"
    )
)

plt.xlabel("time / ns")
plt.ylabel("counts")
plt.title("Photon arrival time")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

# =====================================================================
# 1. DEFINITION DER NORMIERTEN PDF (Wahrscheinlichkeitsdichte)
# =====================================================================
def scintillation_pdf(t, t0, t_rise, t_fall, f_bg, t_min, t_max):
    """
    Normierte PDF für den ungebinnten Fit.
    f_bg ist der Anteil des Untergrunds (Wert zwischen 0 und 1).
    t_min und t_max definieren den Fit-Bereich (wichtig für die Normierung!).
    """
    dt = t - t0
    signal = np.zeros_like(t, dtype=float)
    mask = (dt >= 0) & (t >= t_min) & (t <= t_max)
    
    # Das reine Signal-Modell (ohne Amplitude, da PDF normiert sein muss)
    signal[mask] = (1.0 - np.exp(-dt[mask] / t_rise)) * np.exp(-dt[mask] / t_fall)
    
    # Analytisches Integral des Signals zur korrekten Normierung
    # (Verhindert, dass der Fit die Parameter ins Unendliche schiebt)
    t_start = max(t0, t_min)
    if t_start < t_max:
        # Integral von exp(-dt/t_fall) - exp(-dt*(1/t_rise + 1/t_fall))
        int_fall = t_fall * (1.0 - np.exp(-(t_max - t_start) / t_fall))
        t_comb = 1.0 / (1.0 / t_rise + 1.0 / t_fall)
        int_comb = t_comb * (1.0 - np.exp(-(t_max - t_start) / t_comb))
        signal_integral = int_fall - int_comb
    else:
        signal_integral = 1e-10 # Schutz vor Division durch 0

    # Signal normieren (Fläche = 1)
    signal_pdf = signal / signal_integral
    
    # Flacher Hintergrund normiert auf den Bereich [t_min, t_max]
    background_pdf = 1.0 / (t_max - t_min)
    
    # Kombinierte PDF: Anteil Signal + Anteil Hintergrund
    return (1.0 - f_bg) * signal_pdf + f_bg * background_pdf

# =====================================================================
# 2. DIE NEGATIVE LOG-LIKELIHOOD-FUNKTION (NLL)
# =====================================================================
def negative_log_likelihood(params, data, t_min, t_max):
    t0, t_rise, t_fall, f_bg = params
    
    # Physikalische Grenzen (Strafe/Penalty, falls der Fit in illegale Bereiche läuft)
    if t_rise <= 0 or t_fall <= 0 or f_bg < 0 or f_bg > 1 or t0 < t_min:
        return 1e10
        
    pdf_values = scintillation_pdf(data, t0, t_rise, t_fall, f_bg, t_min, t_max)
    
    # Schutz vor log(0)
    pdf_values = np.clip(pdf_values, 1e-10, None)
    
    # Summe der Logarithmen (negativ, da 'minimize' minimiert, wir aber maximieren wollen)
    return -np.sum(np.log(pdf_values))


rohdaten = top_data["TimeSinceZero"].dropna().to_numpy()

# =====================================================================
# 3. BEISPIEL-DATEN GENERIEREN (Zum Testen)
# =====================================================================
np.random.seed(42)
t_min, t_max = 0, 10
echte_parameter = [10.0, 3.0, 15.0, 0.15] # t0, t_rise, t_fall, f_bg

# Akzeptanz-Verwerfungs-Methode um Testdaten aus der PDF zu ziehen
test_t = np.linspace(t_min, t_max, 10000)
pdf_true = scintillation_pdf(test_t, *echte_parameter, t_min, t_max)
max_pdf = np.max(pdf_true)



# =====================================================================
# 4. FIT AUSFÜHREN
# =====================================================================
# Startwerte für den Fit (grobe Schätzung)
start_werte = [8.0, 1.0, 10.0, 0.05] 

ergebnis = minimize(
    negative_log_likelihood, 
    start_werte, 
    args=(rohdaten, t_min, t_max),
    method='Nelder-Mead' # Robust gegen Stufen, ideal für den Anfang
)

fit_t0, fit_t_rise, fit_t_fall, fit_f_bg = ergebnis.x

print("Fit-Ergebnisse:")
print(f"Startzeit t0:     {fit_t0:.2f} (Echt: {echte_parameter[0]})")
print(f"Anstiegszeit tau: {fit_t_rise:.2f} (Echt: {echte_parameter[1]})")
print(f"Abfallzeit tau:   {fit_t_fall:.2f} (Echt: {echte_parameter[2]})")
print(f"Hintergrund-Anteil: {fit_f_bg*100:.1f}% (Echt: {echte_parameter[3]*100}%)")

# =====================================================================
# 5. PLOT
# =====================================================================
plt.hist(rohdaten, bins=50, density=True, alpha=0.5, color='gray', label='Messdaten (Events)')
x_plot = np.linspace(t_min, t_max, 500)
plt.plot(x_plot, scintillation_pdf(x_plot, fit_t0, fit_t_rise, fit_t_fall, fit_f_bg, t_min, t_max), 
         'r-', lw=2, label='Ungebinnter ML-Fit')
plt.xlabel('Zeit $t$')
plt.ylabel('Normierte Häufigkeit')
plt.legend()
plt.show()
```
