```python
PLOTS_DIR = Path("plots")
PLOTS_DIR.mkdir(exist_ok=True)


def save_plot(filename=None):
    figure = plt.gcf()
    if filename is None:
        title = figure.axes[0].get_title() if figure.axes else "plot"
        filename = re.sub(
            r"[^a-zA-Z0-9_-]+",
            "_",
            title
        ).strip("_").lower()
        filename = filename or "plot"

    figure.savefig(
        PLOTS_DIR / f"{filename}.png",
        dpi=300,
        bbox_inches="tight"
    )
```

```python
def scintillation_pluse_shape(t, A, t0, tau_rise, tau_fall, background):
    t = np.asarray(t, dtype=float)
    dt = t - t0
    shape = np.zeros_like(dt)
    mask = dt > 0
    shape[mask] = (1.0 - np.exp(-dt[mask] / tau_rise)) * np.exp(-dt[mask] / tau_fall)
    return A * shape + background
```

```python
# Fit-Funktion für einen Datensatz
def fit_dataset(
    times,
    bins=1000,
    fit_range=(0, 10)
):

    counts, bin_edges = np.histogram(
        times,
        bins=bins,
        range=fit_range
    )

    bin_centers = 0.5 * (
        bin_edges[:-1] + bin_edges[1:]
    )

    # Poisson errors
    sigma = np.sqrt(
        np.maximum(counts, 1)
    )

    # amplitude, t0, t_rise, t_fall, background
    initial_values = [
        counts.max(),
        0.1,
        0.350,
        1.600,
        0.0
    ]

    bounds = (
        [
            0.0,       # amplitude
            -1.0,      # t0
            0.001,     # rise
            0.05,      # fall
            0.0        # background
        ],
        [
            np.inf,
            2.0,
            10.0,
            20.0,
            np.inf
        ]
    )

    fit_parameters, covariance = curve_fit(
        scintillation_pluse_shape,
        bin_centers,
        counts,
        p0=initial_values,
        sigma=sigma,
        absolute_sigma=True,
        bounds=bounds,
        maxfev=50000
    )

    errors = np.sqrt(
        np.diag(covariance)
    )

    return (
        counts,
        bin_edges,
        bin_centers,
        fit_parameters,
        errors
    )
```

```python
def fit_data(times, bins, fit_range = (0,10)):

    (
        counts_single_event,
        edges_single_event,
        centers_single_event,
        params_single_event,
        errors_single_event
    ) = fit_dataset(
        times,
        bins=bins,
        fit_range=fit_range
    )

    return params_single_event, errors_single_event
```

```python
scint = data[
    data["Process"] == "Scintillation"
].copy()

direct = scint[
    scint["NSideReflections"] == 0
]

reflected = scint[
    scint["NSideReflections"] > 0
]

plt.figure(figsize=(9, 5))

plt.hist(
    direct["TimeSinceZero"],
    bins=500,
    range=(0, 5),
    histtype="step",
    label="direct"
)

plt.hist(
    reflected["TimeSinceZero"],
    bins=500,
    range=(0, 5),
    histtype="step",
    label="reflected"
)

plt.xlabel("Arrival time / ns")
plt.ylabel("Counts")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
fit_bins = 500
fit_range = (0, 5)

fit_sets = {
    "direct": (direct["TimeSinceZero"].dropna().to_numpy(), "tab:blue"),
    "reflected": (reflected["TimeSinceZero"].dropna().to_numpy(), "tab:orange"),
}

param_names = ["A", "t0 / ns", "t_rise / ns", "t_fall / ns", "background"]

plt.figure(figsize=(9, 5))

for label, (times, color) in fit_sets.items():
    counts, edges, centers, params, errors = fit_dataset(
        times,
        bins=fit_bins,
        fit_range=fit_range
    )

    plt.step(
        centers,
        counts,
        where="mid",
        color=color,
        alpha=0.6,
        label=f"{label} (n = {len(times):,})"
    )

    fit_t = np.linspace(*fit_range, 2000)
    plt.plot(
        fit_t,
        scintillation_pluse_shape(fit_t, *params),
        color=color,
        linestyle="--",
        linewidth=2,
        label=(
            rf"{label} fit: $t_{{rise}}={params[2]:.3f}$ ns, "
            rf"$t_{{fall}}={params[3]:.3f}$ ns"
        )
    )

    print(f"\n{label}:")
    for name, value, error in zip(param_names, params, errors):
        print(f"  {name:12s} = {value:.4f} +- {error:.4f}")

plt.xlabel("Arrival time / ns")
plt.ylabel("Counts")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
